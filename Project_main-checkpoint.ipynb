Setup and load data

```python
import numpy as np, pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_percentage_error

LAM = 0.25
END = 2034
Fg  = ['h', 'd_pop', 'd_gdp']

raw = pd.read_csv(r'C:\Users\pc\Desktop\global_p\global_electricity_access_economic_indicatorss.csv')
print(raw.shape)   # (7381, 9)
```

Data cleaning

```python
df = raw.copy()

# Aggregates (World, High income, regions) hatao
kw = ['income','World','IDA','IBRD','OECD','excluding','Euro area','European Union',
      'Small states','Fragile','indebted','Least developed','Arab World','Sub-Saharan',
      'Latin America','East Asia','South Asia','Middle East','North America','dividend',
      'Central Europe','Europe & Central Asia','Oceania']
extra = ['Europe','Eastern Europe','Western Asia','Southern Asia','South Eastern Asia',
         'Not classified','Northern America','Nothern America','Caucasian and Central Asia',
         'Caucasus and Central Asia','Eastern Asia (including Japan)',
         'Eastern Asia (not including Japan)','Africa Eastern and Southern',
         'Africa Western and Central','Caribbean small states','Pacific island small states',
         'Other small states']
mask = df['country'].str.contains('|'.join(kw), case=False, na=False) | df['country'].isin(extra)
df = df[~mask].copy()

# Duplicate naam ek karo
rename = {'Cape Verde':'Cabo Verde','Czech Republic':'Czechia','Turkey':'Turkiye',
          'Vietnam':'Viet Nam',"Korea, Dem. People's Rep.":'Korea, Dem. Rep.',
          'Faeroe Islands':'Faroe Islands','Macedonia, FYR':'North Macedonia',
          'Swaziland':'Eswatini','Somalia, Fed. Rep.':'Somalia','Puerto Rico (US)':'Puerto Rico'}
df['country'] = df['country'].replace(rename)
df = df.groupby(['country','year_recorded'], as_index=False).first()

# Jin territories ki population kabhi nahi mili
no_pop = df.groupby('country')['Total_Population'].apply(lambda s: s.notna().sum())
df = df[~df['country'].isin(no_pop[no_pop == 0].index)].copy()

print(df.shape, df['country'].nunique())   # 216 countries
```

Modeling data

```python
m = df[df['Total_Electricity_Output_GWh'] > 0].dropna(
        subset=['Total_Population', 'GDP_Per_Capita_Current_USD']).copy()

m['log_pop'] = np.log(m['Total_Population'])
m['log_gdp'] = np.log(m['GDP_Per_Capita_Current_USD'])
m['y']       = np.log(m['Total_Electricity_Output_GWh'])

print(m.shape, m['year_recorded'].min(), m['year_recorded'].max())   # (3277, 13), 2000, 2015
```

Helper functions

```python
def make_pairs(d):
    a = d[['country','year_recorded','y','log_pop','log_gdp']]
    p = a.merge(a, on='country', suffixes=('','_0'))
    p = p[p['year_recorded'] > p['year_recorded_0']].copy()
    p['h']     = p['year_recorded'] - p['year_recorded_0']
    p['d_pop'] = p['log_pop'] - p['log_pop_0']
    p['d_gdp'] = p['log_gdp'] - p['log_gdp_0']
    p['d_y']   = p['y'] - p['y_0']
    return p

def drifts(d, G):
    r = {}
    for c, g in d.groupby('country'):
        if len(g) < 6: continue
        sy = np.polyfit(g['year_recorded'], g['y'], 1)[0]
        sp = np.polyfit(g['year_recorded'], g['log_pop'], 1)[0]
        sg = np.polyfit(g['year_recorded'], g['log_gdp'], 1)[0]
        r[c] = sy - (G.coef_[0] + G.coef_[1]*sp + G.coef_[2]*sg)
    return pd.Series(r)

def trend(col, lo=None, hi=None):
    rows = []
    h = df.dropna(subset=[col])
    h = h[(h['year_recorded'] >= 2015) & (h[col] > 0)]
    for c, g in h.groupby('country'):
        if len(g) < 5: continue
        s = np.polyfit(g['year_recorded'], np.log(g[col]), 1)[0]
        if lo is not None: s = float(np.clip(s, lo, hi))
        last = g.sort_values('year_recorded').iloc[-1]
        rows.append((c, np.log(last[col]), s, int(last['year_recorded'])))
    return pd.DataFrame(rows, columns=['country','anchor','slope','yr']).set_index('country')
```

Backtest (train 2000-2011, test 2012-2015)

```python
train = m[m['year_recorded'] <= 2011]
test  = m[(m['year_recorded'] >= 2012) & m['country'].isin(train['country'])]

tp = make_pairs(train)
G2 = LinearRegression().fit(tp[Fg], tp['d_y'])
dr = drifts(train, G2)

tt = test.merge(train[train['year_recorded']==2011][['country','y','log_pop','log_gdp']],
                on='country', suffixes=('','_0'))
tt['h']     = tt['year_recorded'] - 2011
tt['d_pop'] = tt['log_pop'] - tt['log_pop_0']
tt['d_gdp'] = tt['log_gdp'] - tt['log_gdp_0']
tt['dr']    = tt['country'].map(dr).fillna(0)
base_pred   = tt['y_0'] + G2.predict(tt[Fg])

actual = np.exp(tt['y'])
print('Naive baseline MAPE: %.2f%%' % (mean_absolute_percentage_error(actual, np.exp(tt['y_0']))*100))
for lam in [0, 0.25, 0.5, 0.75, 1.0]:
    pr = base_pred + lam * tt['dr'] * tt['h']
    print(f"lam={lam:.2f}  MAPE={mean_absolute_percentage_error(actual, np.exp(pr))*100:.2f}%")
```

Final growth model (poore data pe)

```python
pp = make_pairs(m)
G = LinearRegression().fit(pp[Fg], pp['d_y'])
dr_full = drifts(m, G).clip(-0.05, 0.05)
print(dict(zip(Fg, G.coef_.round(3))))
```

Forecast 2025-2034

```python
base = m[m['year_recorded']==2015].set_index('country')[['y','log_pop','log_gdp']]
base.columns = ['y_0','log_pop_0','log_gdp_0']
P = trend('Total_Population')
D = trend('GDP_Per_Capita_Current_USD', 0, 0.05)

out = []
for c in base.index.intersection(P.index):
    for yr in range(2025, END + 1):
        lp = P.loc[c,'anchor'] + P.loc[c,'slope'] * (yr - P.loc[c,'yr'])
        if c in D.index:
            lg = D.loc[c,'anchor'] + D.loc[c,'slope'] * (yr - D.loc[c,'yr'])
            d_gdp = lg - base.loc[c,'log_gdp_0']
        else:
            d_gdp = 0
        X = pd.DataFrame([[yr - 2015, lp - base.loc[c,'log_pop_0'], d_gdp]], columns=Fg)
        ly = base.loc[c,'y_0'] + G.predict(X)[0] + LAM * dr_full.get(c, 0) * (yr - 2015)
        out.append((c, yr, np.exp(lp), np.exp(ly)))

forecast = pd.DataFrame(out, columns=['country','year','population_pred','electricity_GWh_pred'])
forecast[forecast['country'].isin(['India','Pakistan'])].groupby('country').tail(1)
```
